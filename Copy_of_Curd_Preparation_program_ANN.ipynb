```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Training Data
# [Milk Quality, Temperature, Starter Quantity]

X = np.array([
    [9, 32, 2],
    [8, 30, 2],
    [7, 35, 3],
    [4, 20, 1],
    [5, 22, 1],
    [3, 28, 1],
    [10, 33, 2],
    [9, 34, 3]
])

# Output
# 1 = Good Curd
# 0 = Bad Curd

y = np.array([
    1,
    1,
    1,
    0,
    0,
    0,
    1,
    1
])

# Create ANN Model
model = Sequential()

# Hidden Layer
model.add(Dense(
    units=6,
    activation='relu',
    input_dim=3
))

# Output Layer
model.add(Dense(
    units=1,
    activation='sigmoid'
))

# Compile Model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train Model
model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)

# New Curd Preparation Conditions
new_data = np.array([[8, 31, 2]])

prediction = model.predict(new_data)

print("Probability:", prediction[0][0])

if prediction[0][0] > 0.5:
    print("Good Curd")
else:
    print("Bad Curd")
```
